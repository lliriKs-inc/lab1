# Задача 1
Написать функцию на вход которой подается строка, состоящая из латинских букв.
Функция должна вернуть количество гласных букв (a, e, i, o, u) в этой строке. Написать тесты для кода

```python
def vowels_count(s):
    vowels = "aeiou"
    return sum(1 for char in s.lower() if char in vowels)

assert vowels_count("hello") == 2    # e, o
assert vowels_count("WORLD") == 1    # o
assert vowels_count("Python") == 1   # o
assert vowels_count("AEIOU") == 5    # все гласные
assert vowels_count("bcdfg") == 0    # без гласных

print("Все тесты пройдены успешно")
```

# Задача 2
Написать функцию на вход, которой подается строка. Функция должна вернуть true, если
каждый символ в строке встречается только 1 раз, иначе должна вернуть false. Написать тесты для кода

```python
def all_unique_chars(s):
    return len(set(s)) == len(s)

assert all_unique_chars("abcde") == True    # все уникальные
assert all_unique_chars("hello") == False   # l повторяется
assert all_unique_chars("") == True         # пустая строка, уникальна по определению
assert all_unique_chars("123") == True      # уникальные символы
assert all_unique_chars("1123") == False    # '1' повторяется

print("Все тесты пройдены успешно")
```

# Задача 3
Написать функцию, которая принимает положительное число и возвращает количество
бит равных 1 в этом числе. Написать тесты для кода

```python
def bits_ones_count(n):
    return bin(n)[2:].count('1')

assert bits_ones_count(0) == 0          # двоично 0
assert bits_ones_count(1) == 1          # двоично 1
assert bits_ones_count(7) == 3          # двоично 111
assert bits_ones_count(8) == 1          # двоично 1000
assert bits_ones_count(15) == 4         # двоично 1111

print("Все тесты пройдены успешно")
```

# Задача 4
Написать функцию, которая принимает положительное число. Функция должна вернуть то,
сколько раз необходимо перемножать цифры числа или результат перемножения, чтобы
получилось число состоящее из одной цифры.
Например, для входного числа:
· 39 функция должна вернуть 3, так как 3*9=27 => 2*7=14 => 1*4=4
· 4 функция должна вернуть 0, так как число уже состоит из одной цифры
· 999 функция должна вернуть 4, так как 9*9*9=729 => 7*2*9=126 => 1*2*6=12 =>
1*2=2. Написать тесты для кода

```python
def magic(n):
    count = 0
    while n >= 10:
        result = 1
        for digit in str(n):
            result *= int(digit)
        n = result
        count += 1
    return count

assert magic(39) == 3     # 3*9=27->2*7=14->1*4=4
assert magic(4) == 0      # уже однозначное число
assert magic(999) == 4    # 9*9*9=729->7*2*9=126->1*2*6=12->1*2=2
assert magic(25) == 2     # 2*5=10->1*0=0
assert magic(1) == 0      # однозначное число

print("Все тесты пройдены успешно")
```

# Задача 5
Написать функцию, которая принимает два целочисленных вектора одинаковой длины и
возвращает среднеквадратическое отклонение двух векторов. Написать тесты для кода

```python
def mse(pred, true):
    mse_value = sum((p - t) ** 2 for p, t in zip(pred, true)) / 3
    return mse_value

assert abs(mse([1, 2, 3], [1, 2, 3]) - 0) < 1e-9    # одинаковые векторы, разница 0
assert abs(mse([1, 2, 3], [4, 5, 6]) - 9) < 1e-9    # ((1-4)^2+(2-5)^2+(3-6)^2)/3 = 9
assert abs(mse([0, 0, 0], [1, 1, 1]) - 1) < 1e-9    # ((0-1)^2*3)/3 = 1

print("Все тесты пройдены успешно")
```

# Задача 6
Написать функцию, принимающая целое положительное число. Функция должна вернуть
строку вида “(n1**p1)(n2**p2)…(nk**pk)” представляющая разложение числа на простые
множители (если pi == 1, то выводить только ni).
Например, для числа 86240 функция должна вернуть “(2**5)(5)(7**2)(11)”. Написать тесты для кода

```python
def prime_factors(n):
    factors = {}
    divisor = 2
    while divisor * divisor <= n:
        while n % divisor == 0:
            factors[divisor] = factors.get(divisor, 0) + 1
            n //= divisor
        divisor += 1
    if n > 1:
        factors[n] = factors.get(n, 0) + 1
    
    result = ""
    for prime in sorted(factors):
        power = factors[prime]
        if power == 1:
            result += f"({prime})"
        else:
            result += f"({prime}**{power})"
    return result

assert prime_factors(86240) == "(2**5)(5)(7**2)(11)"
assert prime_factors(12) == "(2**2)(3)"
assert prime_factors(13) == "(13)"
assert prime_factors(100) == "(2**2)(5**2)"
assert prime_factors(1) == ""  # у числа 1 нет простых множителей

print("Все тесты пройдены успешно")
```

# Задача 7
Написать функцию, принимающая целое число n, задающее количество кубиков. Функция
должна определить, можно ли из данного кол-ва кубиков построить пирамиду, то есть
можно ли представить число n как 1^2+2^2+3^2+…+k^2. Если можно, то функция должна
вернуть k, иначе строку “It is impossible”. Написать тесты для кода

```python
def pyramid(number):
    total = 0
    k = 0
    while total < number:
        k += 1
        total += k * k
    if total == number:
        return k
    else:
        return "It is impossible"

assert pyramid(14) == 3                 # 1 + 4 + 9 = 14
assert pyramid(30) == 4                 # 1^2 + 2^2 + 3^2 + 4^2 = 30
assert pyramid(1) == 1                  # 1^2 = 1
assert pyramid(6) == "It is impossible" # невозможно представить
assert pyramid(5) == 2                  # 1^2 + 2^2 = 5
assert pyramid(5 + 9) == 3              # 1+4+9=14, 14 - тест дополнительный

print("Все тесты пройдены успешно")
```

# Задача 8
Функция принимает на вход положительное число и определяет является ли оно сбалансированным, т.е. сумма цифр до средних равна сумме цифр после. Средними в случае нечетного числа цифр считать одну цифру, в случае четного - две средних. Написать тесты для кода

```python
def is_balanced(number):
    s = str(number)
    length = len(s)
    
    if length % 2 == 0:
        left_part = s[:length//2 - 1]
        right_part = s[length//2 + 1:]
    else:
        left_part = s[:length//2]
        right_part = s[length//2 + 1:]
    
    sum_left = sum(int(d) for d in left_part)
    sum_right = sum(int(d) for d in right_part)
    
    return sum_left == sum_right

assert is_balanced(123803) == True    # 1+2+3 == 0+3+0
assert is_balanced(1238031) == False  # 1+2+3 != 0+3+1
assert is_balanced(1) == True          # одно число, сбалансировано по определению
assert is_balanced(12321) == True      # 1+2 == 2+1
assert is_balanced(123456) == False

print("Все тесты пройдены успешно")
```
